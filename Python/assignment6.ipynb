```python
# 1. Wap to extract all the string values present in list only if the string is palindrome. 

L = ['madam', 'hello', 'level', 'python', 'radar']
out = []

for i in L:
    if i == i[::-1]:
        out.append(i)

print(out)
```

```python
# 2. Wap to get the following output.                   S=’hai hello ‘                   Out={‘hai’:’ai’,’hello:’eo’} 

S = 'hai hello'
out = {}

for i in S.split():
    v = ''

    for j in i:
        if j in 'aeiou':
            v += j

    out[i] = v

print(out)
```

```python
# 3. Wap to get the following output.                    S=[‘jiocinema.com’,’file.py’,’web.html’,’amazom.com’,’www.org’]                    Out=[‘com’,’py’,’html’,’org’] 

S = ['jiocinema.com', 'file.py', 'web.html',
     'amazon.com', 'www.org']

out = []

for i in S:
    out.append(i.split('.')[-1])

print(out)
```

```python
# 4. Wap to get the following output.                 S=[‘jiocinema.com’,’file.py’,’web.html’,’amazom.com’,’www.org ’python.py’]                 Out={‘com’:[‘jiocinema’,’amazon’],’py’:[‘file’,’python’],’html’:[‘web’],                       ’org’:[‘www’]} 

S = ['jiocinema.com', 'file.py', 'web.html',
     'amazon.com', 'www.org', 'python.py']

out = {}

for i in S:
    name, ext = i.split('.')

    if ext not in out:
        out[ext] = []

    out[ext].append(name)

print(out)
```

```python
# 5. Wap to get the following output.                  L=[‘hai’,34,3.4,’hello’,90,’byebye’]                  Out={‘hai’:’hi’,’hello’:’ho’,’byebye’:’be’} 
L = ['hai', 34, 3.4, 'hello', 90, 'byebye']
out = {}

for i in L:
    if type(i) == str:
        out[i] = i[0] + i[-1]

print(out)
```

```python
# 6. Wap to check whether the number is neon number or not. 
n = int(input('Enter a number: '))
square = n * n
total = 0

while square > 0:
    digit = square % 10
    total += digit
    square //= 10

if total == n:
    print('Neon number')
else:
    print('Not a Neon number')
```

```python
# 7. Wap to get the following output.                S=’power star’                Out={‘power’:2,’star’:1} 

S = 'power star'
out = {}

for i in S.split():
    count = 0

    for j in i:
        if j in 'aeiou':
            count += 1

    out[i] = count

print(out)
```

```python
# 8. Wap to get the following output.                  In=[100,200,35,40,60]                 Out=[335,235,400,395,375] 
In = [100, 200, 35, 40, 60]
out = []
total = sum(In)

for i in In:
    out.append(total - i)

print(out)
```

```python
# 9. Wap to get the following output.                    In={10:’star’,20:’bye’,30:’moon’,40:’apple’}                    Out={10:’a’,20:’e’,30:’oo’,40:’ae’} 
In = {10: 'star', 20: 'bye', 30: 'moon', 40: 'apple'}
out = {}

for k, v in In.items():
    vowels = ''

    for i in v:
        if i in 'aeiou':
            vowels += i

    out[k] = vowels

print(out)
```

```python
# 10. Wap to get the following output.                     In=[‘hello’,227,3.4,’last’,189,34]                     Out=[722,981,43] 
In = ['hello', 227, 3.4, 'last', 189, 34]
out = []

for i in In:
    if type(i) == int:
        rev = 0

        while i > 0:
            digit = i % 10
            rev = rev * 10 + digit
            i //= 10

        out.append(rev)

print(out)
```
