8. A student scored 78, 85, 92, 88 in four subjects. Calculate the total and average marks.

```python
score = [78, 85, 92, 88]
total = sum(score)
average = total / len(score)
print("Total score:", total)
print("Average score:", average)
```

10. Given: length = 15, breadth = 8. Calculate the area and perimeter of the rectangle. 

```python
length = 15
breadth = 8

area = length * breadth
perimeter = 2 * (length + breadth)

print("Area =", area)
print("Perimeter =", perimeter)
```
