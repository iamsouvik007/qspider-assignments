```python
# 121. WAP to check whether the number is strong or not.

def is_strong(n):
    temp = n
    total = 0

    if n <= 0:
        return False

    while temp > 0:
        digit = temp % 10
        fact = 1

        for i in range(1, digit + 1):
            fact *= i

        total += fact
        temp //= 10

    return total == n


num = int(input("Enter a number: "))

if is_strong(num):
    print("Strong Number")
else:
    print("Not a Strong Number")
```

```python
# 124. WAP to check whether the number is armstrong or not.

def is_armstrong(n):
    temp = n
    digits = len(str(n))
    total = 0

    if n < 0:
        return False

    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10

    return total == n


num = int(input("Enter a number: "))

if is_armstrong(num):
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# 127. WAP to check whether the number is perfect or not.

def is_perfect(n):
    if n <= 1:
        return False

    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


num = int(input("Enter a number: "))

if is_perfect(num):
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
# 130. WAP to check whether the number is prime or not.

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not a Prime Number")
```
