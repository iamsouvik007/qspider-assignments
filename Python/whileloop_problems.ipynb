```python
#  Q50.Write a program to extract all the upper case character ,lower case character ,numbers and special characters into four different output variables from the given string


s = input("Enter the string: ")

upper = ""
lower = ""
numbers = ""
special = ""

i = 0

while i < len(s):
    ch = s[i]

    if ch.isupper():
        upper += ch
    elif ch.islower():
        lower += ch
    elif ch.isdigit():
        numbers += ch
    elif not ch.isspace():
        special += ch

    i += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Numbers:", numbers)
print("Special characters:", special)
```

```python
# Q73.Program to find the union, intersection, and difference of two sets using while loop 


set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

a = list(set1)
b = list(set2)

union = []
intersection = []
difference = []

i = 0

while i < len(a):
    if a[i] not in union:
        union.append(a[i])

    if a[i] in set2:
        intersection.append(a[i])
    else:
        difference.append(a[i])

    i += 1

i = 0

while i < len(b):
    if b[i] not in union:
        union.append(b[i])
    i += 1

print("Union =", set(union))
print("Intersection =", set(intersection))
print("Difference (set1 - set2) =", set(difference))
```

```python
# Q81..Write a program to check weather the given number is spy number or not i.e, 1*2*3=1+2+3 

n = abs(int(input("Enter a number: ")))

temp = n
digit_sum = 0
digit_product = 1

while temp > 0:
    digit = temp % 10
    digit_sum += digit
    digit_product *= digit
    temp //= 10

if digit_sum == digit_product:
    print("Spy number")
else:
    print("Not a Spy number")

```

```python
# Q82.Write a program to check weather the given number is Xylem number or not i.e, 1234 → 1+4=2+3 

n = abs(int(input("Enter a number: ")))
s = str(n)

if len(s) < 3:
    print("Not a Xylem number")
else:
    outer_sum = int(s[0]) + int(s[-1])
    inner_sum = 0
    i = 1

    while i < len(s) - 1:
        inner_sum += int(s[i])
        i += 1

    if outer_sum == inner_sum:
        print("Xylem number")
    else:
        print("Not a Xylem number")

```

```python
# Q83.Write a program to check weather the given number is phloem number or not I.e, 12345→ 1+5!=2+3+4 

n = abs(int(input("Enter a number: ")))
s = str(n)

if len(s) < 3:
    print("Not a Phloem number")
else:
    outer_sum = int(s[0]) + int(s[-1])
    inner_sum = 0
    i = 1

    while i < len(s) - 1:
        inner_sum += int(s[i])
        i += 1

    if outer_sum != inner_sum:
        print("Phloem number")
    else:
        print("Not a Phloem number")

```

```python
#  Q84.Write a program to check weather the given number is neon number or not i.e. 9 is number, 9**2=81→8+1=9 


n = int(input("Enter a number: "))

square = n * n
digit_sum = 0

while square > 0:
    digit_sum += square % 10
    square //= 10

if n >= 0 and digit_sum == n:
    print("Neon number")
else:
    print("Not a Neon number")
```

```python
# Q85.Write a program to check weather the given number is automorphic or not I.e.5 is number 5**2=25 last digit of 25 is the number itself 


n = int(input("Enter a number: "))

square = n * n

if str(square).endswith(str(n)) and n >= 0:
    print("Automorphic number")
else:
    print("Not an Automorphic number")
```

```python
# Q88.Write a program to count number of consonants in the given string s=input('enter the str:') 

s = input("Enter the string: ")

count = 0
i = 0

while i < len(s):
    ch = s[i].lower()

    if ch.isalpha() and ch not in "aeiou":
        count += 1

    i += 1

print("Number of consonants =", count)

```

```python
# Q91.Write a program to check weather the given collection is having nested collection or not 


items = eval(input("Enter a collection: "))

i = 0
found = False

while i < len(items):
    if isinstance(items[i], (list, tuple, set, dict)):
        found = True
        break
    i += 1

if found:
    print("Nested collection found")
else:
    print("No nested collection")
```

```python
# Q92. Write a program to return the positions of vowels present in the given string 

s = input("Enter the string: ")

positions = []
i = 0

while i < len(s):
    if s[i].lower() in "aeiou":
        positions.append(i)
    i += 1

print("Vowel positions =", positions)

```

```python
# Q97.Write a program to get the following output - Input='hai hello how are you' output='hai**hello**how**are**you' 

s = input("Enter the string: ")

words = s.split()
result = ""
i = 0

while i < len(words):
    result += words[i]

    if i < len(words) - 1:
        result += "**"

    i += 1

print(result)

```

```python
# Q98.Write a program to reverse the given list 
items = eval(input("Enter a list: "))

reversed_items = []
i = len(items) - 1

while i >= 0:
    reversed_items.append(items[i])
    i -= 1

print("Reversed list:", reversed_items)
```

```python
# Q99.Write a program for number game 

import random 
num = random.randint(1, 100)
attempts = 0
while True:
    guess = int(input("Guess the number between 1 and 100: "))
    attempts += 1
    if guess < num:
        print("Too low! Try again.")
    elif guess > num:
        print("Too high! Try again.")
    else:
        print("Congratulations! You guessed the number in", attempts, "attempts.")
        break
```
